### Text Tokenization & Feature Extraction

- Tokenization prepares text for downstream tasks like stopword removal, stemming, and vectorization. It’s the first step in converting unstructured text into structured data.
- Once text is tokenized, Feature Extraction converts it into numerical form so ML models can process it.

```python
# TEXT CLEANING FUNCTIONS

stopwords_set = set(stopwords.words('english'))

def tokenize_text(text):
    text = str(text)
    words = word_tokenize(text.lower())
    words = [w for w in words if w not in string.punctuation and w not in stopwords_set]
    return list(set(words))  # return unique words

nltk.download('punkt')
nltk.download('punkt_tab')   # <— add this line
nltk.download('stopwords')

# Apply tokenization
df['tokenized_side_effects'] = df['side_effects'].apply(tokenize_text)
df['tokenized_medical_desc'] = df['medical_condition_description'].apply(tokenize_text)

# Related Drugs Extraction (Improved)
def extract_related_drugs(text):
    s = str(text)
    if s.lower() in ('nan', 'none', '', 'unknown'):
        return []
    matches = re.findall(r'([^:\|]+):', s)
    cleaned = [m.strip() for m in matches if m.strip()]
    return cleaned

df['tokenized_related_drugs'] = df['related_drugs'].apply(extract_related_drugs)

print("✅ Preprocessing completed successfully!")
print(df[['tokenized_side_effects', 'tokenized_related_drugs']].head())
```

### Side Effect Co-occurrence Heatmap

```python
# Create a binary matrix: drugs x side effects
all_side_effects = list(set([item for sublist in df['tokenized_side_effects'] for item in sublist]))
side_effect_matrix = pd.DataFrame(0, index=df['drug_name'], columns=all_side_effects)
```

```python
for idx, row in df.iterrows():
    side_effect_matrix.loc[row['drug_name'], row['tokenized_side_effects']] = 1
```

```python
from itertools import combinations
from collections import Counter
```

```python
# Create pairwise counts
pair_counts = Counter()

for effects in df['tokenized_side_effects']:
    unique_effects = sorted(set(effects))
    for combo in combinations(unique_effects, 2):
        pair_counts[combo] += 1

# Convert to DataFrame
co_occurrence_df = pd.DataFrame(
    pair_counts.items(),
    columns=['pair', 'count']
)
co_occurrence_df[['effect1', 'effect2']] = pd.DataFrame(co_occurrence_df['pair'].tolist(), index=co_occurrence_df.index)
co_occurrence_df.drop(columns='pair', inplace=True)
```

```python
# Create pivot table for visualization (Top 20)
top_effects = co_occurrence_df.groupby('effect1')['count'].sum().nlargest(20).index
filtered = co_occurrence_df[
    co_occurrence_df['effect1'].isin(top_effects) &
    co_occurrence_df['effect2'].isin(top_effects)
]

heatmap_data = filtered.pivot_table(
    index='effect1', columns='effect2', values='count', fill_value=0
)
```
